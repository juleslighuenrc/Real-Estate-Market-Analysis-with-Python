```python
dfcust.columns = dfcust.columns.str.strip().str.replace('\ufeff', '')
```

```python
dfcust['customerid']
```

```python

dfcust.columns.values
```

```python
dfcust = dfcust.drop(['Unnamed: 0'], axis = 1)
```

```python
dfcust.columns.values
```

```python
dfcust['full_name'] = dfcust['name'] + ' ' + dfcust['surname']
dfcust.head()
```

```python
dfcust = dfcust.drop(columns=['name', 'surname'])
dfcust
```

```python
dfcust.dtypes
```

```python
dfcust['entity'].unique()
```

```python
entity_bool_columns = pd.get_dummies(dfcust['entity'], drop_first = True, dtype = int)
entity_bool_columns
```

```python
dfcust['entity'] = entity_bool_columns.iloc[:, 0]
dfcust.head()
```

```python
dfcust['mortgage'].unique()
```

```python
dfcust['mortgage'] = pd.get_dummies(dfcust['mortgage'], drop_first=True, dtype=int)
dfcust.head(3)
```

```python
dfcust['country'].unique()
```

```python
dfcust['country'].value_counts()
```

```python
dfcust['country'] = dfcust['country'].str.strip()
dfcust['country'].value_counts()
```

```python
dfcust['country'] = np.where(dfcust['country'] == 'USA', 1, 0)
dfcust['country'].unique()
```

```python
dfcust.info()
```

```python
dfcust['state'].unique()
```

```python
dfcust['state'].value_counts()
```

```python
dfcust['state'] = dfcust['state'].fillna('Other')
dfcust['state'] = (dfcust['state'] == 'California').astype(int)
dfcust.head()
```

```python
dfcust['state'].value_counts()
```

```python
dfcust.info()
```

```python
dfcust['birth_date'] = pd.to_datetime(dfcust['birth_date'], errors='coerce')
dfcust['birth_date'].head()
```

```python
today = pd.Timestamp.today()
dfcust['age'] = (today - dfcust['birth_date']).dt.days / 365.25
dfcust['age'].head()
```

```python
dfcust['age'] = dfcust['age'].fillna(dfcust['age'].median())
dfcust['age'] = dfcust['age'].astype(int)
dfcust.head()
```

```python
dfcust = dfcust.drop(columns=['birth_date'])
dfcust_cleaned = dfcust.copy()
```

```python
dfcust_cleaned.head()
```

```python
dfcust_cleaned.info()
```

```python
dfcust.columns.values
```

```python
dfcust_columns_reordered = ['customerid', 'full_name', 'age', 'entity', 'sex', 'country',
                            'state', 'purpose',
                            'deal_satisfaction', 'mortgage', 'source']
dfcust_cleaned = dfcust_cleaned[dfcust_columns_reordered]
dfcust_cleaned.head()
```

```python
dfcust_cleaned['purpose'].value_counts()
```

```python
dfcust_cleaned['purpose'] = np.where(dfcust_cleaned['purpose'] == 'Home', 1, 0)
dfcust_cleaned['purpose'].unique()
```

```python
dfcust_cleaned.head()
```

```python
dfcust_cleaned['sex'] = dfcust_cleaned['sex'].fillna('Unknown')
dfcust_cleaned['sex'].value_counts()
```
