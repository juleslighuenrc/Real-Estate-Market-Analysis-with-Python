```python
years_since_sale = (pd.Timestamp.today() - pd.to_datetime(dfmerged['date_sale'])).dt.days / 365.25
years_since_sale
```

```python
dfmerged1['age_at_sale'] = dfmerged1['age'] - years_since_sale.round()
dfmerged1.head()
```

```python
dfmerged1['age_at_sale'].min()
```

```python
dfmerged1['age_at_sale'].max()
```

```python
df_sold = dfmerged1[dfmerged1['status'] == 1]
df_sold['price_bin'] = pd.cut(dfmerged1['price'], bins = 10)
df_sold['age_bin'] = pd.cut(dfmerged1['age_at_sale'],bins = 10)
df_sold
```

```python
cum_age = df_sold['age_bin'].value_counts().sort_index().cumsum()
cum_price = df_sold['price_bin'].value_counts().sort_index().cumsum()
cum_age
```

```python
cum_price
```

```python
freq_table = pd.crosstab(df_sold['age_bin'],df_sold['price_bin'])
cum_freq_by_price = freq_table.cumsum(axis=0)
cum_freq_by_price
```

```python
corr = df_sold[['age_at_sale','price']].corr()
corr
#The Pearson correlation coefficient between age at sale and property price is -0.155, 
#indicating a very weak negative linear relationship. This suggests that as age increases, 
#property prices tend to decrease slightly, but the relationship is not strong.
```

```python
cov = df_sold[['age_at_sale','price']].cov()
cov
```

```python
#confirming the negative relation between age and price: as age increases,
#prices decrease. Perhaps, this is due to taking less risks at an older age,
#a change of priorities, or more informed decisions as to what it is needed.
```

```python
#However, these results are tricky: the cumulative frequency table by price shows that middle-age groups
#buy more properties(therefore they are the main clientele), whereas younger groups or older
#tend to buy less properties. This tells us that withing each age group, purchases accumulate
#gradually across price bins, meaning that more expensive properties are more common
#among middle-aged buyers. 
```
